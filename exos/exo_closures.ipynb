# Exercices
## Cache 1 valeur

````python
import time

def long_call():

    previous = (None, None)

    def inner_call(value:int):
        nonlocal previous

        previous_value, previous_result = previous
        if value == previous_value:
            result = previous_result
        else:
            time.sleep(2)
            result = value**2
            previous = (value, result)

        return result

    return inner_call

long_call_instance = long_call()
````

````python
print(long_call_instance(5))
print(long_call_instance(3))
print(long_call_instance(3))
````

## Cache N valeurs

````python
import time
from collections import OrderedDict

def long_call():

    previous = OrderedDict()

    def inner_call(value:int):
        if value in previous:
            result = previous[value]
            previous.move_to_end(value)
        else:
            time.sleep(2)
            result = value**2

            previous[value] = result

            if len(previous) > 3:
                v = previous.popitem(last=False)
                print(v)

        return result

    return inner_call

long_call_instance = long_call()
````

````python
print(long_call_instance(1))
print(long_call_instance(2))
print(long_call_instance(1))
print(long_call_instance(3))
print(long_call_instance(4))
````
