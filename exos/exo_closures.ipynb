# Exercice closures

Fonction de base :
```python
import time

def long_call(value:int):
    time.sleep(2)
    return value**2
```

````python
import time

def long_call():

    previous = (None, None)

    def inner_call(value:int):
        nonlocal previous

        previous_value, previous_result = previous
        if value == previous_value:
            result = previous_result
        else:
            time.sleep(2)
            result = value**2
            previous = (value, result)

        return result

    return inner_call

long_call_instance = long_call()
````

````python
print(long_call_instance(5))
print(long_call_instance(3))
print(long_call_instance(3))
````
