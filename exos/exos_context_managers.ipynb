# Les context managers

## Exercice 1

```python
import time

class TimeIt():
    def __enter__(self):
        self.start = time.time()

    def __exit__(self, exc_type, exc_val, exc_tb):
        end = time.time()
        print(f"{end - self.start:.4f}secondes se sont écoulées")
        del self.start

timer = TimeIt()

with timer:
    for x in range(1_000_000):
        y = x ** 2
```

## Second exercice

```python
class IndentContext:
    def __init__(self):
        self._intent_count = 0

    def __enter__(self):
        self._intent_count += 1
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        self._intent_count -= 1

    def print(self, text:str):
        print("\t" * self._intent_count, text, sep="")
```

```python
print("Start")

with IndentContext() as ic:
    ic.print("une ligne")
    ic.print("une autre ligne")
    with ic:
        ic.print("une ligne indentée")
    ic.print("une autre ligne")

ic.print("out")

print("end")
```

## Troisième exerices

```python
import os

class ChangeDir:
    def __init__(self, target_dir):
        self._target_dir = target_dir

    def __enter__(self):
        self._origin_dir = os.getcwd()
        os.chdir(self._target_dir)
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        os.chdir(self._origin_dir)
        del self._origin_dir

print(os.getcwd())

with ChangeDir('/System') as chdir_obj:
    print(os.getcwd())

print(os.getcwd())
```
