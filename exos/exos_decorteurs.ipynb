````python
import time

def timeit(func):
    def wrapper(*args, **kwargs):
        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()
        print("Elapsed time: ", end - start)
        return result
    return wrapper
````

````python
@timeit
def longrun():
    for x in range(1_000_000):
        y = x ** 2
````

````python
longrun()
````

## Second exercice

````python
def count_calls(func):
    count = 0

    def inner(*args, **kwargs):
        result = func(*args, **kwargs)
        nonlocal count
        count += 1

        return result

    def inner_get_count():
        return count

    inner.get_count = inner_get_count

    return inner
````

````python
@count_calls
@timeit
def some_func():
    print("func called")
````

````python
some_func()
some_func()
some_func()
````

````python
some_func.get_count()
````

## Troisième exercice

````python
import time

def long_call(func):

    previous = (None, None)

    def inner_call(value:int):
        nonlocal previous

        previous_value, previous_result = previous
        if value == previous_value:
            result = previous_result
        else:
            result = func(value)
            previous = (value, result)

        return result

    return inner_call

@long_call
def long_func(value:int):
    time.sleep(2)
    return value**2

long_call_instance = long_call()
````

## Quatrième exercice

````python
channels = []

def register(level):
    def inner_register(func):
        channels.append((level, func))
        return func
    return inner_register

@register(level=2)
def notify_mail():
    print("notification on mail")

@register(level=3)
def notify_sms():
    print("notification on message")

@register(level=0)
def notify_push():
    print("notification on push service")

def send_notifications(level):
    for ch_level, channel in channels:
        if ch_level > level:
            channel()
````

````python
send_notifications(1) # mail et push appelés
````
