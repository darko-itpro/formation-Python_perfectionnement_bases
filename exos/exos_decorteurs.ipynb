````python
import time

def timeit(func):
    def wrapper(*args, **kwargs):
        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()
        print("Elapsed time: ", end - start)
        return result
    return wrapper
````

````python
@timeit
def longrun():
    for x in range(1_000_000):
        y = x ** 2
````

````python
longrun()
````
