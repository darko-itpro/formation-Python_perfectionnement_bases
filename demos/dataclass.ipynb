# Les dataclasses

```python
import dataclasses as dc

@dc.dataclass(frozen=True)
class Point:
    x: int
    y: int
    z: int = None
    color: list = dc.field(default=None, compare=False)


    def distance_to(self, point: Point) -> float:
        print("in distance calculation")
```

```python
p1 = Point(4, 8, color=["red"])
p2 = Point(4, 8, color=["blue"])
```

```python
p1 == p2
```

```python
p1.x = 10
```

```python
p2 = Point(4, 8)
```

```python
p1 == p2
```
