## Les générateurs

```python
(name.upper() for name in camelot)
```

```python
for i in (name.upper() for name in camelot):
    print(i)
```

```python
def upperize(kingdom:list):
    for name in kingdom:
        yield name.upper()

    yield "toto"
    print('Done')
```

```python
upperize(camelot)
```

```python
for i in upperize(camelot):
    print("-", i)
```

```python
def fibo(max=5000000000000):
    value = 1
    previous = 1

    while value < max:
        yield value
        value, previous = value + previous, value
```

```python
for i in fibo():
    print(i)
```
