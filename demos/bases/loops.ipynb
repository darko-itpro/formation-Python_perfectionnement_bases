# Boucles et itérations

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
jobs = ["king", "wizard", "knight", "knight", "knight", "knight",]
```

## Itérer sur plusieurs itérables avec `zip()`

```python
for name, job in zip(camelot, jobs):
    print(name, job)
```

## Les lists comprehension

```python
new_list = []
for name in camelot:
    new_list.append(name.upper())
```

```python
[name.upper() for name in camelot]
```

```python
[len(name) for name in camelot]
```

```python
camelot = [("Arthur", "king"), ("Merlin", "wizard"),
           ("Lancelot", "knight"), ("Robin", "knight"),
           ("Perceval", "knight"), ("Karadoc", "knight")]
```

```python
[name for name, job in camelot if job != "wizard"]
```

```python
[name.upper() 
 for name, job in camelot 
 if job != "wizard"]
```

```python
[peon for peon in camelot if peon[1] != "wizard"]
```

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
jobs = ["king", "wizard", "knight", "knight", "knight", "knight",]
```

```python
[peon for peon in zip(camelot, jobs)]
```

```python
[(name, job) for name, job in zip(camelot, jobs)]
```

```python
[(name, job) for name, job in zip(camelot, jobs) if job != "wizard"]
```

```python
{name: job 
 for name, job in zip(camelot, jobs)}
```

```python
import pylib.datasources.pyacademy as pa
```

```python
for item in zip(pa.load_students_list(),
    pa.load_grades("math"),
    pa.load_grades("english")):
    print(item)
```

```python
{name: (math, english)
for name, math, english in zip(pa.load_students_list(),
    pa.load_grades("math"),
    pa.load_grades("english"))}
```

```python
{name: {"math":math, "english":english}
for name, math, english in zip(pa.load_students_list(),
    pa.load_grades("math"),
    pa.load_grades("english"))}
```
