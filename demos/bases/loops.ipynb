# Boucles et itérations

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
jobs = ["king", "wizard", "knight", "knight", "knight", "knight",]
```

## Itérer sur plusieurs itérables avec `zip()`

```python
for name, job in zip(camelot, jobs):
    print(name, job)
```

```python
[name for name, job in camelot if job != "wizard"]
```

```python
[name.upper() 
 for name, job in camelot 
 if job != "wizard"]
```

```python
[peon for peon in camelot if peon[1] != "wizard"]
```

## Générateurs

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
jobs = ["king", "wizard", "knight", "knight", "knight", "knight",]
```

```python
for item in (name.upper() for name in camelot):
    print(item)
```

```python
def upperise(kingdom):
    for name in kingdom:
        yield name.upper()

    print("end")
```

```python
for item in upperise(camelot):
    print("from for", item)
```

```python
my_gen = upperise(camelot)
```

```python
next(my_gen)
```

```python
def repas():
    yield 'Entrée'
    yield 'Plat'
    yield 'dessert'
    yield 'café'
```

```python
ce_midi = repas()
```

```python
next(ce_midi)
```

```python
def tick():
    count = 0
    while True:
        count += 1
        yield count
```

```python
t = tick()
```

```python
next(t)
```
