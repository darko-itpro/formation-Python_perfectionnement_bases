# Rappels sur l'objet

```python
class Training:
    def __init__(self, subject:str, duration:int, seats:int=12):
        self.subject = subject
        self.duration = duration
        self._students = []
        
    def add_student(self, name):
        self._students.append(name)
```

```python
t = Training("Python", 3, 9)
```

```python
t.add_student('Elliot Alderson')
```

```python
t._students
```

```python
t.add_student
```

```python
Training.add_student(t,"Elliot")
```
