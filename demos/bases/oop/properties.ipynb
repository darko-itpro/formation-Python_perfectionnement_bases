```python
def extract_students(self):
    return self._students

def give_name(self):
    return self._name

def update_name(self, new_name):
    self._name = new_name.title()
```

```python
class Training:
    def __init__(self, name):
        self._name = name
        self._students = []

    students = property(fget=extract_students)
    name = property(fget=give_name, fset=update_name)
```

```python
t = Training("Python")
```

```python
t.students
```
