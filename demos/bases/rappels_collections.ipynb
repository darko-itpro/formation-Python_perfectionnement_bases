## Packing et Unpacking

```python
current_time = 1156
```

```python
current_time // 60
```

```python
"Il est {}h{}".format(current_time // 60, current_time % 60)
```

```python
"Il est {}h{}".format(*divmod(current_time, 60))
```

## Variadics

```python
def func(mi_id, *args):
    print(args)
```

```python
func("id", 3)
```

```python
func(mi_id="id", param = 42)
```

```python
def func(**kwargs):
    print(kwargs)
```

```python
func(param=2)
```

```python
def func(*args, **kargs):
    pass
```

```python
help(print)
```

```python
params = (2, 3, 4)
```

```python
def func(a, b, c):
    print(a, b, c)
```

```python
func(*params)
```

```python
paramsd = {'a': 5, 'b': 3, 'c': 10}
```

```python
func(**paramsd)
```
