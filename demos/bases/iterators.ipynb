```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
(name.upper() for name in camelot)   
```

```python
for i in (name.upper() for name in camelot)   :
    print(i)
```

```python
def upperize(kingdom):
    for name in kingdom:
        yield name.upper()

    print("done")

for i in upperize(camelot):
    print(i)
```

```python
ci = upperize(camelot)
```

```python
next(ci)
```

```python
def lunch():
    yield "entrée"
    yield "plat"
    yield "formage"
    yield "dessert"
    yield "café"
```

```python
l = lunch()
```

```python
next(l)
```

```python
def count():
    c = 0
    while True:
        c += 1
        yield c
```

```python
c = count()
```

```python
next(c)
```
