# Les Itérateurs


```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
class KingdomIterator:
    def __init__(self, kingdom):
        self._kingdom = kingdom
        self._curindex = None

    def __iter__(self):
        self._curindex = 0
        return self

    def __next__(self):
        try:
            item = self._kingdom[self._curindex]
            self._curindex += 1
            return item
        except IndexError:
            raise StopIteration
```

```python
k = KingdomIterator(camelot)
k_iterator = k.__iter__()
```

```python
next(k_iterator)
```

```python
registery_iterator = k.__iter__()
```

```python
next(registery_iterator)
```
