# Les Itérateurs


```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
class Kingdom:
    def __init__(self, peons):
        self._peons = peons

    def __iter__(self):
        self._cursor = 0
        return self

    def __next__(self):
        try:
            item = self._peons[self._cursor]
            self._cursor += 1
            return item
        except IndexError:
            raise StopIteration
```

```python
k = Kingdom(camelot)
for i in k:
    print(i)
```
