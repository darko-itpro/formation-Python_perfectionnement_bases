```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
class MyClass:
    def __init__(self, peons:list):
        self.students = peons
    
    def __iter__(self):
        self._current_value = 0
        return self

    def __next__(self):
        try:
            value = self.students[self._current_value]
            self._current_value += 1
            return value
        except IndexError:
            raise StopIteration("End of collection")
```

```python
mc = MyClass(camelot)
```

```python
for peon in mc:
    print(peon)
```

```python
iter_mc = mc.__iter__()
```

```python
next(iter_mc)
```

```python
dir(mc)
```
