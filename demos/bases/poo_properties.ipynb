```python
class Training:
    def __init__(self, subject, duration, seats= 12):
        self.subject = subject.title()
        self.duration = duration
        self.seats = seats
        self._students = []

    @property
    def remaining_seats(self):
        return self.seats - len(self._students)

    @property
    def students(self):
        return self._students.copy()

    @property
    def subject(self):
        return self.subject

    @subject.getter
    def subject(self, new_subject):
        self._subject = new_subject.title()
    

    def add_student(self, name:str):
        self._students.append(name)
```

```python
t = Training("Python", 4)
t.add_student("Johno Doe")
```

## Real Implementation

```python
def get_students(obj):
    return obj._students.copy()

def del_students(self):
    self._students.clear()

def get_subject(obj):
    return obj._subject

def set_subject(obj, new_subject):
    obj._subject = new_subject.title()
```

```python
class Training:    
    def __init__(self, subject, duration, seats= 12):
        self._subject = subject.title()
        self.duration = duration
        self.seats = seats
        self._students = []

    students = property(get_students, fdel=del_students)

    subject = property(get_subject, set_subject)

    def add_student(self, name:str):
        self._students.append(name)
```

```python
t = Training("Django", 4)
t.add_student("John")
```

```python
t.students
```

```python
t.students = []
```
