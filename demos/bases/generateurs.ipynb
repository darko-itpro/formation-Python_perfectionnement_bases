```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
[len(knight) for knight in camelot]
```

```python
camelot = [("Arthur", "king"), ("Merlin", "wizard"),
           ("Lancelot", "knight"), ("Robin", "knight"),
           ("Perceval", "knight"), ("Karadoc", "knight")]
```

```python
[name for name, job in camelot if job != "wizard"]
```

```python
for item in (name for name, job in camelot if job != "wizard") :
    print(item)
```

```python
def send_to_quest(kingdom_registery):
    for name, job in kingdom_registery:
        if job != "wizard":
            yield name

    print("done")
```

```python
for item in send_to_quest(camelot):
    print(item)
```

```python
quest_group = send_to_quest(camelot)
```

```python
next(quest_group)
```

```python
def menu():
    yield "entrée"
    yield "plat"
    yield "fromage"
    yield "dessert"
    yield "café"
```

```python
m = menu()
```

```python
next(m)
```

```python
def counter():
    count = 0
    while True:
        count += 1
        yield count
```

```python
c = counter()
```

```python
next(c)
```
