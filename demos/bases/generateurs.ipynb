# Les générateurs

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
upperized = (name.upper() for name in camelot)
```

```python
type(upperized)
```

```python
upperized
```

```python
for item in (name.upper() for name in camelot):
    print(item)
```

```python
def upperize(kingdom: list):
    for item in kingdom:
        yield item.upper()
    print('Done')
```

```python
for item in upperize(camelot):
    print(item)
```

```python
gen = upperize(camelot)
```

```python
next(gen)
```

```python
def menu():
    yield "entrée"
    yield "Plat"
    yield "dessert"
    yield "café"
```

```python
m = menu()
```

```python
next(m)
```

```python
def count():
    c = 0
    while True:
        c +=1
        yield c
```

```python
counter = count()
```

```python
next(counter)
```

```python
def receiver():
    print("Setting up")
    while True:
        data = yield
        yield data * 2
```

```python
r = receiver()
```

```python
next(r)
```
