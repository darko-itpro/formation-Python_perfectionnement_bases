```python
class Training:
    def __init__(self, subject:str, duration:int, seats: int = 12):
        self.subject = subject
        self._students = []
        self.duration = duration
        self.seats = seats

    @property
    def subject(self):
        return self._subject

    @subject.setter
    def subject(self, new_name:str):
        self._subject = new_name.title()

    @property
    def available_seats(self):
        return self.seats - len(self._students)

    @property
    def students(self):
        return self._students.copy()

    def add_student(self, name):
        if name in self._students:
            raise ValueError('Duplicate student')
        
        self.students.append(name)


t = Training("Python", 5)
print(t.subject)
t.students
print(t.available_seats)
```

```python
def give_students(self):
    return self._students.copy()

def set_students(obj, students:list):
    obj._students = students

```

```python
class Training:
    
    def __init__(self, subject:str, duration:int, seats: int = 12):
        self.subject = subject.title()
        self._students = []
        self.duration = duration
        self.seats = seats

    def _reset_students(self):
        self._students.clear()
    
    students = property(fget=give_students, fset=set_students, fdel=_reset_students)

    def add_student(self, name):
        if name in self._students:
            raise ValueError('Duplicate student')
        
        self.students.append(name)
```

```python
t = Training("Python", 5)
t.students
```

```python
t = Training("python", 5)
```

```python
t.students = ["toto"]
```
