# Les properties

```python
def get_students(self):
    return self._students.copy()

def set_students(self, students: list):
    self._students[:] = students[:]

def del_students(self):
    self._students.clear()

class Training:
    def __init__(self, title: str, max_seats: int|str = 10):
        self.title = title.upper()
        self.max_seats = int(max_seats)
        self._students = []

    students = property(fget=get_students, fset=set_students, fdel=del_students)

    def add_student(self, student: str):
        self._students.append(student)
```

```python
python = Training("Python")
```

```python
python.students = []
```

```python
python.add_student
```

```python
python.add_student("toot")
```

```python
python.students
```

## Descripteurs

```python
class Answer:
    def __get__(self, instance, owner):
        return 42

class TurtleAccess:
    def __get__(self, instance, owner):
        return instance._turtle_name

    def __set__(self, instance, value):
        instance._turtle_name = value
```

```python
class Universe:
    def __init__(self, turtle_name: str):
        self._turtle_name = turtle_name

    meaning_of_life = Answer()
    turtle = TurtleAccess()
```

```python
discworld = Universe("A'Tuin")
```

```python
discworld.meaning_of_life
```
