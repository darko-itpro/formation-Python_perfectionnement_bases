```python
def provide_students(self):
    return self._students

def set_students(training, new_students:list):
    training._students[:] = new_students
```

```python
class Training:
    def __init__(self, title, duration, max_seats= 12):
        self.title = title
        self.duration = duration
        self.seats = max_seats
        self._students = []

    students = property(fget=provide_students, fset=set_students)

    
    def add_student(self, name):
        if len(self._students) >= self.seats:
            raise ValueError("Training Full")
            
        self._students.append(name)
```

```python
t = Training("Python", 5)
```

```python
t.students
```
