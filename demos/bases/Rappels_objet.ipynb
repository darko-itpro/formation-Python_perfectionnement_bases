```python
def provide_students(self):
    return self._students

def set_students(training, new_students:list):
    training._students[:] = new_students
```

```python
class Training:
    def __init__(self, title, duration, max_seats= 12):
        self.title = title
        self.duration = duration
        self.seats = max_seats
        self._students = []

    students = property(fget=provide_students, fset=set_students)

    
    def add_student(self, name):
        if len(self._students) >= self.seats:
            raise ValueError("Training Full")
            
        self._students.append(name)
```

## Objet en tant que conteneur

```python
class Training:
    
    def __init__(self, title, duration, max_seats= 12):
        self.title = title
        self.duration = duration
        self.seats = max_seats
        self.students = []
    
    def add_student(self, name):
        if len(self.students) >= self.seats:
            raise ValueError("Training Full")
            
        self.students.append(name)

    def __str__(self):
        return f'Training {self.title}, {len(self.students)} students'

    def __repr__(self):
        return f'Training({self.title}, {self.duration})'

    def __len__(self):
        return len(self.students)

    def __contains__(self, item):
        return item in self.students

    def __getitem__(self, value):
        print(value, type(value))

python = Training('Python', 5)
python.add_student("Eliott Alderson")
python.add_student("Kevin Mitnik")
```
