## Rappels de programmation orientée objet

```python
class Training:
    def __init__(self, name:str, duration:int, max_seats:int=12):
        self.name = name.title()
        self.duration = duration
        self._students = []
        self.max_seats = max_seats

    def students(self) -> list:
        return self._students.copy()

    def my_name(self):
        return self.name

    def add_student(self, name:str):
        self.students.append(name)
```

```python
t = Training("Python avancé", 5)
```

```python
t.my_name()
```

```python
Training.my_name(t)
```

## Properties

```python
def get_x(self):
    return self._x

def set_x(obj, x:int):
    if x < 0:
        raise ValueError("x can't be negative")
    obj._x = x


class Point:
    def __init__(self, x:int, y:int):
        self._x = x
        self._y = y

    x = property(fget=get_x, fset=set_x)
```

```python
p = Point(4, 8)
```

```python
p.x = -1
```
