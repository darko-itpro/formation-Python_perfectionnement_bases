# L'Objet

## Rappels de programmation orientée objet

```python
class Training:
    def __init__(self, name:str, duration:int, max_seats:int=12):
        self.name = name.title()
        self.duration = duration
        self._students = []
        self.max_seats = max_seats

    def students(self) -> list:
        return self._students.copy()

    def my_name(self):
        return self.name

    def add_student(self, name:str):
        self.students.append(name)
```

```python
t = Training("Python avancé", 5)
```

```python
dir(t)
```

```python
t.my_name()
```

```python
Training.my_name(t)
```
