```python
def func(myid, *args):
    print(myid, args)
```

```python
def func(**kwargs):
    print(kwargs)
```

```python
func(value=3, other="toto")
```

```python
def display_time(hours, minutes):
    print("{}h{}".format(hours, minutes))
```

```python
display_time(3, 34)
```

```python
display_time(*divmod(234, 60))
```

```python
def display_time(*, hours, minutes):
    print("{}h{}".format(hours, minutes))
```

```python
display_time(hours=3, minutes=34)
```

```python
duration = {"hours": 3, "minutes":34}
```

```python
display_time(**duration)
```
