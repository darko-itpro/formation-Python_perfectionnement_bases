```python
class Training:
    def __init__(self, subject: str, duration:int, max_seats:int=12):
        self.subject = subject
        self.duration = int(duration)
        self.max_seats = int(max_seats)
        self.students = []

    def add_student(self, name):
        self.students.append(name)

    def mystudents(self):
        return self.students
```

### Des fonctions pour les properties

```python
def give_turtle_name(obj):
    return obj._turtle_name
```

```python
def set_turtle_name(obj, name):
    obj._turtle_name = name
```

```python
class DiscWorld:
    def __init__(self):
        self._turtle_name = None

    turtle_name = property(give_turtle_name, set_turtle_name)
```

```python
dw = DiscWorld()
```

```python
dw.turtle_name = "A'Tuin"
```

```python
dw.turtle_name
```

```python
dw = DiscWorld()
set_turtle_name(dw, "A'Tuin")
print(give_turtle_name(dw))
```
