# Les générateurs

```python
for i in range(10):
    print(i)
```

```python
camelot = ["Arthur", "Merlin", "Lancelot", "Robin", "Perceval", "Karadoc"]
```

```python
[len(knight) for knight in camelot]
```

```python
gen = (len(knight) for knight in camelot)
```

```python
for i in gen:
    print(i)
```

```python
def getsizes(kingdom):
    for knight in kingdom:
        yield len(knight)

    print("done")
```

```python
for i in getsizes(camelot):
    print(i)
```

```python
gen = getsizes(camelot)
```

```python
next(gen)
```

```python
def dinner():
    yield "entrée"
    yield "plat"
    yield "dessert"
    yield "café"
```

```python
tonight = dinner()
```

```python
next(tonight)
```

```python
def counter():
    count = 0
    while True:
        count += 1
        yield count
```

```python
c = counter()
```

```python
next(c)
```
